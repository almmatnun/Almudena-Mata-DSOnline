Una matriz recoge las horas semanales de 3 clases extraescolares durante 4 semanas. Calcula totales por clase y por semana con axis, localiza la clase con más horas (argmax) y dibújalo.

````python
import numpy as np
import matplotlib.pyplot as plt

horas = np.array([[3, 4, 3, 5],
                  [2, 2, 3, 2],
                  [4, 5, 4, 6]])
clases = ["CAI", "ER", "NMPC"]

print("Total general:", horas.sum())
print("Horas por clase (axis=1):", horas.sum(axis=1))
print("Horas por semana (axis=0):", horas.sum(axis=0))

idx = horas.sum(axis=1).argmax()
print("Clase con mas horas:", clases[idx])

plt.figure(figsize=(7, 4))
plt.bar(clases, horas.sum(axis=1), color="#4d9fff")
plt.title("Horas totales por clase extraescolar")
plt.ylabel("horas")
plt.tight_layout()
plt.show()
````

Un slice de un array es una vista: comparte memoria con el original, así que modificarlo cambia también el original. .copy() rompe ese vínculo.

Ojo: si escribes bloque = bloque * 10, Python crea un array nuevo con el resultado y lo guarda en bloque, así que la vista se pierde y original no cambia. Repite el experimento con un segundo array de los mismos valores (por ejemplo original2), esta vez rompiendo el vínculo antes de multiplicar, y comprueba que ese segundo array no cambia.

Rotar un vector 90° alrededor del eje Z es girarlo en sentido antihorario visto desde arriba: la altura z no cambia y lo que gira es el plano xy. Lo vas a hacer con una matriz 3×3 y lo que viste arriba: cada columna de la matriz dice adónde va a parar un eje.

Cómo leer el código de partida: v es el vector que vas a girar. theta es el ángulo, 90°, pasado a radianes con np.deg2rad porque NumPy mide los ángulos así. El corrector lo usa para saber qué giro esperar, así que déjalo en 90. El dibujo del final ya está escrito: pinta v en azul y v_rot en rojo.

Cómo abordarlo, paso a paso:

En un papel, dibuja los ejes x e y vistos desde arriba y gíralos 90° en sentido antihorario. ¿Dónde acaba la punta del eje x, [1, 0, 0]? ¿Y la del eje y, [0, 1, 0]? ¿Y el eje z, [0, 0, 1]?
Esos tres destinos son las tres columnas de la matriz. Guárdala en Rz.
Aplica Rz al vector con el producto matricial y guarda el resultado en v_rot.
Imprime v, v_rot y las dos normas, y ejecuta.
Tiene que salir algo así como esto (si tu matriz lleva decimales, los números saldrán con punto, como -1.):

v original: [2 1 1]
v rotado:   [-1  2  1]
norma antes: 2.45 -> norma despues: 2.45
La norma no cambia porque rotar no estira el vector. Eso no basta para saber que tu matriz gira: un espejo también conserva la longitud.

Para ir más allá: con cualquier ángulo θ, el eje x va a parar a [cos θ, sin θ, 0] y el eje y a [-sin θ, cos θ, 0]; en NumPy, np.cos(theta) y np.sin(theta). Cuando lo tengas con 90°, prueba a escribir así las columnas y cambia el ángulo a 45: compara los números con el preajuste «Rotación Z 45°» del laboratorio de arriba.

````python
import numpy as np
import matplotlib.pyplot as plt

# RETO: rota el vector 3D v 90 grados alrededor del eje Z
v = np.array([2, 1, 1])

theta = np.deg2rad(90)                 # el angulo, en radianes; dejalo en 90 para el corrector

# Cada columna es adonde va un eje al girar 90 grados (antihorario, visto desde arriba):
#   columna 0: el eje x [1, 0, 0] pasa a [0, 1, 0]
#   columna 1: el eje y [0, 1, 0] pasa a [-1, 0, 0]
#   columna 2: el eje z [0, 0, 1] no se mueve
Rz = np.array([[0, -1, 0],
               [1,  0, 0],
               [0,  0, 1]])
# Con cualquier angulo, las mismas columnas en forma general:
# Rz = np.array([[np.cos(theta), -np.sin(theta), 0],
#                [np.sin(theta),  np.cos(theta), 0],
#                [0,              0,             1]])
v_rot = Rz @ v                         # aplica la matriz al vector

print("v original:", v)
print("v rotado:  ", np.round(v_rot, 2))
print("norma antes:", round(np.linalg.norm(v), 2),
      "-> norma despues:", round(np.linalg.norm(v_rot), 2))

# El dibujo ya esta escrito: funciona en cuanto existan v y v_rot
fig = plt.figure(figsize=(5, 5))
ax = fig.add_subplot(111, projection="3d")
ax.quiver(0, 0, 0, v[0], v[1], v[2], color="#4d9fff", label="v")
ax.quiver(0, 0, 0, v_rot[0], v_rot[1], v_rot[2], color="#dd0b16", label="v rotado")
lim = 3
ax.set_xlim(-lim, lim)
ax.set_ylim(-lim, lim)
ax.set_zlim(-lim, lim)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
ax.legend()
plt.show()
````
