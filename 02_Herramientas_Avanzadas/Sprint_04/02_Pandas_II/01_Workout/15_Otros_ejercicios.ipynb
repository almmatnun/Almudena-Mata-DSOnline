Seleccionar el subconjunto de filas que cumple una o varias condiciones. Con Series se usa &/|, nunca and/or.

Los datos: seis vuelos. Créalos tú en la consola, en un DataFrame llamado df con index=["V1", "V2", "V3", "V4", "V5", "V6"] y estas columnas:

compania = ["PamPangea", "AirBridge", "PamPangea", "AirBridge", "PamPangea", "AirBridge"]
destino = ["Ginebra", "Melbourne", "Ginebra", "Melbourne", "NY", "Ginebra"]
distancia = [900, 16000, 950, 15500, 5800, 1100]
Qué tienes que hacer (imprime cada resultado con print, que es lo que lee la corrección):

Imprimir los vuelos de PamPangea (máscara sobre compania).
Imprimir los de más de 10000 km con destino Melbourne (& y paréntesis).
Imprimir los de PamPangea o con destino Ginebra (| y paréntesis).

```python
import numpy as np
import pandas as pd

s_compania = pd.Series(["PamPangea", "AirBridge", "PamPangea", "AirBridge", "PamPangea", "AirBridge"], index = ["V1", "V2", "V3", "V4", "V5", "V6"])

s_destino = pd.Series(["Ginebra", "Melbourne", "Ginebra", "Melbourne", "NY", "Ginebra"], index = ["V1", "V2", "V3", "V4", "V5", "V6"])

s_distancia = pd.Series([900, 16000, 950, 15500, 5800, 1100], index = ["V1", "V2", "V3", "V4", "V5", "V6"])

df = pd.DataFrame({"compania": s_compania, "destino": s_destino, "distancia": s_distancia})

print(df)


#Imprimir los vuelos de PamPangea (máscara sobre compania).

print(df.loc[df["compania"] == "PamPangea"])

#Imprimir los de más de 10000 km con destino Melbourne (& y paréntesis).
mas_mil = df["distancia"] > 1000
a_Melbourne = df["destino"] == "Melbourne"

condicion = mas_mil & a_Melbourne

print(df.loc[condicion])

#Imprimir los de PamPangea o con destino Ginebra (| y paréntesis).

company = df["compania"] == "PamPangea"

destino_ciudad = df["destino"] == "Ginebra"

print(df.loc[company | destino_ciudad])
```

Usar una máscara con loc para corregir valores solo en las filas que cumplen una condición, y crear una columna booleana derivada.

Los datos: cuatro vuelos. Créalos tú en la consola, en un DataFrame llamado df con estas columnas:

compania = ["TabarAir", "TabarAir", "PamPangea", "AirBridge"]
destino = ["Los Angeles", "Roma", "Los Angeles", "Tokio"]
distancia = [9000, 1200, 15500, 11000]
Qué tienes que hacer:

Los vuelos de TabarAir a "Los Angeles" iban en realidad a "San Francisco": corrige el destino de solo esas filas, con df.loc[mascara, "destino"].
Crear la columna Supertrayecto: True si distancia es mayor que 12000, y False si no.
Mostrar cuántos supertrayectos hay (la suma de esa columna booleana).

```python
df = pd.DataFrame({"compania" : ["TabarAir", "TabarAir", "PamPangea", "AirBridge"], "destino" : ["Los Angeles", "Roma", "Los Angeles", "Tokio"], "distancia" : [9000, 1200, 15500, 11000]})

print(df)


#Los vuelos de TabarAir a "Los Angeles" iban en realidad a "San Francisco": corrige el destino de solo esas filas, con df.loc[mascara, "destino"].

es_TabarAir = df["compania"] == "TabarAir"
a_LosAngeles = df["destino"] == "Los Angeles"

df.loc[es_TabarAir & a_LosAngeles, "destino"] = "San Francisco"

print(df)

#Crear la columna Supertrayecto: True si distancia es mayor que 12000, y False si no.

df["Supertrayecto"] = df["distancia"] > 12000

print(df)


#Mostrar cuántos supertrayectos hay (la suma de esa columna booleana).

print(df["Supertrayecto"].value_counts())
```

Define una función propia y aplícala con apply a cada valor de una columna para crear una columna categórica nueva.

Los datos: cinco vuelos. Créalos tú en la consola, en un DataFrame llamado df con estas columnas:

vuelo = ["V1", "V2", "V3", "V4", "V5"]
distancia = [16000, 1500, 5800, 900, 12000]
Qué tienes que hacer:

Escribir una función clasificador(distancia) que devuelva "LD" si la distancia es mayor que 10000, "MD" si es mayor o igual que 2000, y "CD" en el resto de casos.
Crear la columna df["categoria"] aplicando esa función con apply sobre df["distancia"], e imprimir el DataFrame.

```python
df = pd.DataFrame({"vuelo" : ["V1", "V2", "V3", "V4", "V5"], "distancia" : [16000, 1500, 5800, 900, 12000]})

print(df)

#Escribir una función clasificador(distancia) que devuelva "LD" si la distancia es mayor que 10000, "MD" si es mayor o igual que 2000, y "CD" en el resto de casos.

def clasifica_distancia (distancia):
        if distancia > 10000:
            clasificacion = "LD"
        elif distancia >= 2000:
            clasificacion = "MD"
        else:
            clasificacion = "CD"
        return clasificacion
                      
#Crear la columna df["categoria"] aplicando esa función con apply sobre df["distancia"], e imprimir el DataFrame.

df["categoria"] = df["distancia"].apply(clasifica_distancia)

print(df)
```

Ir más allá de una sola función de agregación: contar por grupo, obtener la moda con una función propia y aplicar distintas funciones por columna con agg y un diccionario.

Los datos: seis vuelos de tres compañías. Créalos tú en la consola, en un DataFrame llamado df con estas columnas:

compania = ["A", "A", "A", "B", "B", "C"]
destino = ["NY", "NY", "Roma", "Bali", "Bali", "Lima"]
consumo = [4000, 4200, 3800, 6000, 6200, 2500]
duracion = [8, 8, 2, 16, 15, 12]
Qué tienes que hacer (imprime cada resultado con print, que es lo que lee la corrección):

Contar el número de vuelos por compañía (groupby + size).
Sacar el destino más visitado por compañía, o sea la moda. Escribe la función def mas_visitado(destinos): return destinos.mode()[0] y pásala con df.groupby("compania")["destino"].apply(mas_visitado): .mode() devuelve los valores más repetidos y [0] se queda con el primero.
Con .agg y un diccionario, aplicar "mean" a consumo y "max" a duracion.

```python
df = pd.DataFrame({"compania" : ["A", "A", "A", "B", "B", "C"], "destino" : ["NY", "NY", "Roma", "Bali", "Bali", "Lima"], "consumo" : [4000, 4200, 3800, 6000, 6200, 2500], "duracion" : [8, 8, 2, 16, 15, 12]})

print(df)

#Contar el número de vuelos por compañía (groupby + size).

print(df.groupby(["compania"]).size())


#Sacar el destino más visitado por compañía, o sea la moda. Escribe la función def mas_visitado(destinos): return destinos.mode()[0] y pásala con df.groupby("compania")["destino"].apply(mas_visitado): .mode() devuelve los valores más repetidos y [0] se queda con el primero.

def mas_visitado(destinos):
    return destinos.mode()[0]

print(df.groupby(["compania"])["destino"].apply(mas_visitado))

#Con .agg y un diccionario, aplicar "mean" a consumo y "max" a duracion.

agregaciones = {"consumo": "mean", "duracion": "max"}

print(df.groupby(["compania"])[["consumo", "duracion"]].agg(agregaciones))
```

A diferencia de agg (una fila por grupo), transform devuelve una serie de la misma longitud que la entrada, ideal para pegar a cada fila el valor de su grupo.

Los datos: seis vuelos de tres compañías. Créalos tú en la consola, en un DataFrame llamado df con estas columnas:

compania = ["A", "A", "B", "B", "B", "C"]
duracion = [8, 10, 16, 14, 15, 12]
Qué tienes que hacer:

Calcular la media de duracion por compañía con groupby + mean (sale una fila por grupo).
Con transform("mean"), pegar a cada fila la media de su compañía, en una columna duracion_media_cia.
Crear desviacion = duracion - duracion_media_cia.

```python
df = pd.DataFrame({"compania" : ["A", "A", "B", "B", "B", "C"], "duracion" : [8, 10, 16, 14, 15, 12]})

print(df)


#Calcular la media de duracion por compañía con groupby + mean (sale una fila por grupo).

print(df.groupby("compania")["duracion"].mean())


#Con transform("mean"), pegar a cada fila la media de su compañía, en una columna duracion_media_cia.

duracion_media_cia = df.groupby("compania")["duracion"].transform("mean")
print(duracion_media_cia)

df['duracion_media_cia'] = df.groupby('compania')['duracion'].transform('mean')

#Crear desviacion = duracion - duracion_media_cia.

df["desviacion"] = df["duracion"] - df["duracion_media_cia"]

print(df)
```
